5. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/
Cloud in terms of accessibility and data security?


-  Tableau Public: Free, but publicly accessible to everyone on the internet.
Tableau Server/Cloud: Paid, private, and secure inside an organization with strict permission controls.



6. When publishing a Tableau dashboard, what happens to a live connection compared to an extract?
How does this affect data refresh and performance?\


- Live Connection: Queries the database directly. Data is always real-time, but performance can be slow if the database is large.

Extract: Takes a snapshot/copy of the data into Tableau's memory. It is much faster and works offline, but requires manual or scheduled data refreshes to get new updates.


7. Explain the core philosophy difference between Power BI and Tableau in terms of reporting and exploration.


- Power BI: Built around Reporting & Integration—it connects easily with Microsoft ecosystems (Excel, Azure) and focuses on structured business reports.

Tableau: Built around Data Exploration & Visual Analytics—it focuses heavily on deep visual data discovery and flexibility.

2 strengths of Tableau

- Power BI:

Strengths: Free with Microsoft tools, great data modeling (DAX).

Weaknesses: Less flexible for custom visual formatting compared to Tableau, tightly bound to the Windows ecosystem.

Tableau:

Strengths: Industry-leading data visualization, handles massive datasets smoothly.

Weaknesses: Expensive licensing, steeper learning curve for advanced features.



9. Define a KPI and explain why KPIs are important in executive reporting.

- Definition: A Key Performance Indicator (KPI) is a measurable value that shows how effectively an organization is achieving key business objectives.

Importance: Executives are busy; KPIs give them a quick, high-level summary of performance at a single glance without needing to look at messy details.
